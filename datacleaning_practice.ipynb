```python
#importing python library
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Step 2: Create Dataset
data = {
    "Name": ["Aman", "Riya", "Rahul", "Neha", "Karan"],
    "Marks": [80, 90, 70, 85, 95]
}

df = pd.DataFrame(data)

print("Student Dataset:")
print(df)
```

```python
#Step 3: Explore Dataset
print("\nFirst 5 Rows:")
print(df.head())

print("\nDataset Shape:")
print(df.shape)

print("\nDataset Information:")
df.info()

print("\nStatistical Information:")
print(df.describe())
```

```python
# Step 4: NumPy Analysis
marks = np.array(df["Marks"])

print("\nAverage Marks:", np.mean(marks))
print("Highest Marks:", np.max(marks))
print("Lowest Marks:", np.min(marks))
```

```python
# Step 5: Data Cleaning
# Creating one missing value
df.loc[2, "Marks"] = np.nan

print("\nMissing Values:")
print(df.isnull().sum())

# Filling missing value with average
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())

print("\nCleaned Dataset:")
print(df)
```

```python
# Step 6: Bar Chart
plt.figure(figsize=(7, 4))
plt.bar(df["Name"], df["Marks"])
plt.title("Student Marks")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.show()
```

```python
# Step 7: Line Chart
plt.figure(figsize=(7, 4))
plt.plot(df["Name"], df["Marks"], marker="o")
plt.title("Student Marks")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.show()
```

```python
# Step 8: Pie Chart
plt.figure(figsize=(7, 7))
plt.pie(df["Marks"], labels=df["Name"], autopct="%1.1f%%")
plt.title("Marks Distribution")
plt.show()
```
